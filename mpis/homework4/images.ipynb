```python
from igraph import *
from math import floor

N = 8
```

```python
g = Graph.Full(N)
g.vs[0]['color'] = 'green'
g.vs['label'] = g.vs.indices
plot(g)
```

```python
g = Graph(N)
g.add_edges([(i, i + 1) for i in range(N - 1)])
g.vs[N // 2]['color'] = 'green'
g.vs['label'] = g.vs.indices
plot(g)
```

```python
g = Graph(N)
g.add_edges([(i, i + 1) for i in range(N - 1)])
g.vs[0]['color'] = 'green'
g.vs['label'] = g.vs.indices
plot(g)
```

```python
g = Graph.Tree(N, 2)
g.vs[0]['color'] = 'green'
g.vs['label'] = g.vs.indices
layout = g.layout_reingold_tilford(root=[0])
plot(g, layout=layout)
```

```python
g = Graph.Full(floor(2 * N / 3))
g.add_vertices(N - floor(2 * N / 3))
g.add_edges([(i, i + 1) for i in range(floor(2 * N / 3) - 1, N - 1)])

g.vs[0]['color'] = 'green'
g.vs['label'] = g.vs.indices
plot(g)
```
